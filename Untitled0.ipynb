```python

# Hospital Assistant Chatbot
# Created by: Rutvi Shah

print("=" * 40)
print("     HOSPITAL ASSISTANT CHATBOT")
print("=" * 40)
print("Welcome to CareWell Hospital!")
print("Ask me about hospital services.")
print("Type 'bye' to exit.")

while True:
    user = input("\nYou: ").lower().strip()

    if user in ["hi", "hello", "hey"]:
        print("Bot: Hello! Welcome to CareWell Hospital.")

    elif "appointment" in user or "doctor" in user:
        print("Bot: Please visit the hospital reception to book an appointment.")

    elif "timing" in user or "hours" in user:
        print("Bot: Sample OPD timings are 9 AM to 6 PM.")

    elif "department" in user:
        print("Bot: We have General Medicine, Cardiology, Pediatrics and Orthopedics.")

    elif "fee" in user or "fees" in user:
        print("Bot: Please contact the hospital reception for consultation fees.")

    elif "emergency" in user or "ambulance" in user:
        print("Bot: For an emergency in India, call 112 or 108 for ambulance assistance.")

    elif "pharmacy" in user or "medicine" in user:
        print("Bot: Please ask the hospital reception about pharmacy services.")

    elif "address" in user or "location" in user:
        print("Bot: Please check the hospital's official address.")

    elif "thank" in user:
        print("Bot: You're welcome! Stay healthy.")

    elif user in ["bye", "exit", "quit"]:
        print("Bot: Thank you for visiting. Goodbye!")
        break

    elif user == "":
        print("Bot: Please enter a question.")

    else:
        print("Bot: Sorry, I did not understand.")
        print("Bot: Ask about appointments, timings, fees or departments.")



```
